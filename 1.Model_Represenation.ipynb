```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("ggplot")
```

```python
x_train = np.array([1,2,3,4])
y_train = np.array([10,15,20,25])
print("x_train: ",x_train)
print("y_train: ",y_train)
```

```python
b = 5
w = 5
```

```python
def compute_plot(x,w,b):

    m = x.shape[0]
    fb= np.zeros(m)
    for i in range(m):
        fb[i] = w*x[i] + b
    return fb
     
```

```python
z = compute_plot(x_train,w,b)
plt.scatter(x_train,y_train,marker= 's',c = 'b',label = 'actual values')
plt.plot(x_train, z,c = 'r',label = 'predicted values' )
plt.title("Housing prediction")
plt.xlabel('Size(in 1000sq.ft)')
plt.ylabel('price(in 100,000$)')
plt.legend()
plt.show()


```

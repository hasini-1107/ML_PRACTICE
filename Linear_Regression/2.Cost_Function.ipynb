```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("classic")
```

```python
x_train = np.array([1,2,3,4])
y_train = np.array([20,40,52,68])
```

```python
w = 16
b = 4
```

```python
def costfunct(x,y,w,b):
    size = x.shape[0]
    fb = np.zeros(size)
    sum = 0
    for i in range(size):
        fb[i]= x[i]*w + b
        error = (fb[i]-y[i])**2
        sum = sum + error
    cost_function = (1/(2*size))*sum
    return cost_function
```

```python
cost = costfunct(x_train,y_train,w,b)
print("cost = ", cost)
```

```python

f_wb = x_train*w + b
plt.scatter(x_train, y_train, marker = 'o',c = 'g',label = "actual values")
plt.plot(x_train,f_wb, c= 'b',label = "predicted values")
plt.title("cost")
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

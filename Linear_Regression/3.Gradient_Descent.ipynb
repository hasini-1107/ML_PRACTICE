```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("classic")

```

```python
x_train = np.array([1,4,8,9])
y_train = np.array([20,45,87,91])
```

```python
w = 0
b = 0
alpha = 0.01
iteration = 10000
```

```python

def gradient_descent(x,y,alpha,w,b):
 for steps in range(iteration):
    m = x.shape[0]
    fb = np.zeros(m)
    dw = 0
    db = 0
    for i in range(m):
        fb[i]=w*x[i]+b
        error = fb[i]-y[i]
        dw = dw + error*x[i]
        db = db + error 
    dw = dw/m
    db = db/m
    w = w - alpha*dw
    b = b - alpha*db
 return w,b
```

```python
w,b = gradient_descent(x_train,y_train,alpha,w,b)
print(w, b)
```

```python
plt.scatter(x_train, y_train, marker = 'o',c = 'g',label = "actual values")
f_wb = w*x_train + b
plt.plot(x_train,f_wb, c= 'b',label = "predicted values")
plt.title("gradient function")
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
